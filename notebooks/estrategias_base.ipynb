```python
import seaborn as sns

tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
from sklearn.model_selection import train_test_split

df = tips.copy()
# Definir atributos preditivos (X) e o alvo (y)
X = df.drop('tip', axis=1) #Tarefa de Regressão
y = df['tip']
print(X.head())
```

```python
print(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
                                                    X,
                                                    y,
                                                    test_size=0.3,
                                                    random_state=42)
print('Linhas de Treino')
print(len(X_train))
print(len(y_train))
print('Linhas de Teste')
print(len(X_test))
print(len(y_test))
```

```python
df = tips.copy()

# Definir atributos preditivos (X) e alvo (y) para classificação
X = df.drop('smoker', axis=1)
y_classe = df['smoker']
print(len(X))
print(len(y))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y_classe,
    test_size = 0.2,
    random_state = 42,
    stratify = y_classe)
print('Treino:')
print(len(X_train))
print(len(y_train))
print(y_train.value_counts(normalize=True).mul(100))
print('Teste')
print(len(X_test))
print(len(y_test))
print(y_test.value_counts(normalize=True).mul(100))
```

```python
#Estratégia de Validação cruzada com Estratificação

from sklearn.model_selection import StratifiedKFold
df = tips.copy()
X = df.drop('smoker', axis=1)
y = df['smoker']

cv_str = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

for i,(treino, teste) in enumerate(cv_str.split(X, y)):
    print(f'Fold {i+1}')
    print(len(treino))
    print(len(teste))
```
